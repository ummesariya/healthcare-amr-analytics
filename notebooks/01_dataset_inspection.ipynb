# Data Dictionary

## 1. Culture Dataset

| Column | Meaning | Data Role |
|---|---|---|
| anon_id | Deidentified patient identifier | Patient identifier |
| pat_enc_csn_id_coded | Deidentified patient encounter number | Encounter identifier |
| order_proc_id_coded | Deidentified unique culture order number | Culture-order identifier |
| order_time_jittered | Jittered timestamp associated with the culture order | Temporal variable |
| ordering_mode | Mode in which the culture was ordered | Clinical/order feature |
| culture_description | Type of culture ordered, such as Blood, Respiratory, or Urine | Culture feature |
| was_positive | Binary indicator of whether the culture result was positive | Culture outcome |
| organism | Name of microorganism identified | Microbiological feature |
| antibiotic | Antibiotic tested | Susceptibility feature |
| susceptibility | Susceptibility result for the tested antibiotic | Resistance outcome |

## Important Dataset Notes

- The dataset is deidentified.
- Missing values are represented by the literal value `"Null"` according to the dataset README.
- Dates and times have been jittered.
- A single culture order can contain multiple rows.
- A single culture order can contain multiple organisms and/or antibiotics.
- Susceptibility is recorded at the antibiotic-testing level.

## 2. Demographics Dataset

| Column | Meaning | Data Role |
|---|---|---|
| anon_id | Deidentified patient identifier | Patient identifier |
| pat_enc_csn_id_coded | Deidentified patient encounter number | Encounter identifier |
| order_proc_id_coded | Deidentified unique culture order number | Culture-order identifier |
| age | Patient age represented as an age group | Demographic feature |
| gender | Binary gender indicator; the dataset does not identify which value represents which gender | Demographic feature |

## Important Dataset Notes

- Age is grouped into categories rather than provided as an exact age.
- Gender is represented by binary indicators.
- The dataset documentation does not identify which binary value corresponds to which gender.
- `"Null"` represents missing information in the source dataset.

## 3. Project-Level Data Dictionary

This section maps the available variables to their role in the healthcare analytics and antimicrobial resistance prediction project.

| Variable | Source Dataset | Project Role | ML Consideration |
|---|---|---|---|
| anon_id | Culture + Demographics | Patient identifier | Identifier; not used directly as a predictor |
| pat_enc_csn_id_coded | Culture + Demographics | Encounter identifier | Identifier; not used directly as a predictor |
| order_proc_id_coded | Culture + Demographics | Culture-order identifier | Identifier; used to define the analysis unit |
| order_time_jittered | Culture | Culture timing | Potential temporal feature; timestamps must respect prediction timing |
| ordering_mode | Culture | Clinical/order feature | Potential predictor |
| culture_description | Culture | Type of microbiological culture | Potential predictor |
| was_positive | Culture | Culture positivity outcome | Analytics variable; potential secondary outcome |
| organism | Culture | Identified microorganism | Requires careful timing assessment before ML use |
| antibiotic | Culture | Antibiotic tested | Context variable for susceptibility analysis |
| susceptibility | Culture | Antibiotic susceptibility result | Primary resistance-related outcome |
| age | Demographics | Patient age group | Potential predictor |
| gender | Demographics | Binary demographic indicator | Potential predictor; coding meaning is not identified |

### Project Data Rules

- Patient, encounter, and culture-order identifiers will not be used as predictive features.
- The susceptibility result represents the resistance-related outcome and must not be used as an input feature when predicting that outcome.
- Variables will only be used as ML predictors if they would be available at the defined prediction point.
- Organism information requires a timing assessment because organism identification may occur during the microbiology workflow.
- The final ML feature set will be determined after the dataset structure and prediction timeline are fully investigated.
- No AMR target definition will be finalized until the susceptibility data and relevant clinical variables have been examined.
